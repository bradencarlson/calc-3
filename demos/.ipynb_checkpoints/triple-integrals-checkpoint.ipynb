```python
x,y,z = var('x y z')

def f(x,y):
    return x*y
    
x_range = (0,1);
y_range = (0,1);
z_range = (0,1);
surf = plot3d(f(x,y), x_range, y_range)
surf = surf.add_condition(lambda x,y,z: y<= x)

bottom = plot3d(0, x_range, y_range)
bottom = bottom.add_condition(lambda x,y,z: y <= x)

side1 = implicit_plot3d(y==x, x_range, y_range, z_range)
side1 = side1.add_condition(lambda x,y,z: 0<=z<=f(x,y))

side2 = implicit_plot3d(x == 1, x_range, y_range, z_range)
side2 = side2.add_condition(lambda x,y,z: 0<=z<=f(x,y))

show(surf + bottom + side1 + side2)
```

```python
x,y,z = var('x y z')
x_range = (x,-3,3);
y_range = (y,-1,5);
z_range = (z,-3,3);
surf = implicit_plot3d( y == x^2 + z^2, x_range, y_range, z_range)
surf = surf.add_condition(lambda x,y,z: y<= 4)

top = implicit_plot3d( y==4, x_range, y_range, z_range)
top = top.add_condition(lambda x,y,z: x^2 + z^2 <= 4)

proj = implicit_plot3d( y == 0, x_range, y_range, z_range, color='green')
proj = proj.add_condition(lambda x,y,z: x^2 + z^2 <= 4)

show( surf + top )
```

```python
x,y,z = var('x y z')

x_r = (x, 0, 2)
y_r = (y,-1,1)
z_r = (z,-1,2)

def f(x,y): 
    return 2 - 2*y-x;

surf = implicit_plot3d(z == f(x,y), x_r, y_r, z_r)
surf = surf.add_condition(lambda x, y, z: x >= 2*y and y>=0 and z>=0)

bottom = implicit_plot3d(  z==0, x_r, y_r, z_r)
bottom = bottom.add_condition(lambda x,y,z: x>=2*y and f(x,y) >= 0 and y>=0)

side = implicit_plot3d( x == 2*y, x_r, y_r, z_r)
side = side.add_condition(lambda x,y,z: 0<=z<=f(x,y))

side1 = implicit_plot3d( y == 0.00, x_r, y_r, z_r)
side1 = side1.add_condition(lambda x,y,z: 0<= z <= f(x,y))

show( surf + bottom + side + side1)
```
